```python
import numpy as np

class AlibabaWorkloadLoader:
    """Processes the Alibaba PAI trace for LRMA simulation."""
    def __init__(self, pods_file, nodes_file):
        self.pods_df = pd.read_csv(pods_file)
        self.nodes_df = pd.read_csv(nodes_file)
        self._preprocess()

    def _preprocess(self):
        # Fill missing GPU specs with a default 'Unknown' or 'MISC'
        self.pods_df['gpu_spec'] = self.pods_df['gpu_spec'].fillna('MISC')

        # Convert time to relative starting points (seconds or normalized slots)
        min_time = self.pods_df['creation_time'].min()
        self.pods_df['start_slot'] = self.pods_df['creation_time'] - min_time
        self.pods_df['duration'] = self.pods_df['deletion_time'] - self.pods_df['creation_time']

    def get_tasks_at_time(self, t_start, t_end):
        """Extract tasks that arrive within a specific time window."""
        mask = (self.pods_df['start_slot'] >= t_start) & (self.pods_df['start_slot'] < t_end)
        return self.pods_df[mask]

# Initialize loader with one of the spec files
loader = AlibabaWorkloadLoader(
    '/content/openb_pod_list_gpuspec33.csv',
    '/content/openb_node_list_gpu_node.csv'
)

print(f"Loaded {len(loader.pods_df)} tasks and {len(loader.nodes_df)} GPU nodes.")
print("Sample processed tasks:")
display(loader.pods_df[['name', 'cpu_milli', 'gpu_milli', 'gpu_spec', 'start_slot', 'duration']].head())
```

### Task Modeling and Environment Constants
We now define the constants for the IoT environment as described in the LRMA paper, including wireless parameters and the core Task structure.

```python
class LRMATask:
    def __init__(self, task_id, arrival_time, cpu_req, gpu_req, gpu_type, duration):
        self.task_id = task_id
        self.t_arrival = arrival_time

        # Properties mapping to paper variables {size, C, G, R}
        self.C = cpu_req / 1000.0  # CPU resource demand (C_i,k)
        self.G = gpu_req / 1000.0  # GPU resource demand (G_i,k)
        self.R = gpu_type          # GPU type requirement (R_i,k)

        # Derived size property (size_i,k) in MB
        # Mapping CPU/GPU intensity and duration to a data payload proxy
        self.size = (self.C * 0.5 + self.G * 2.0) * np.log1p(duration / 1e6)

        self.duration = duration

    def __repr__(self):
        return f"LRMA_Task(ID={self.task_id}, size={self.size:.2f}MB, C={self.C}, G={self.G}, R={self.R})"

# Simulation Constants
class EnvConfig:
    NUM_EDS = 50
    NUM_MES = 5
    V = 10.0

# Demonstration
sample_row = loader.pods_df.iloc[0]
sample_task = LRMATask(
    task_id=sample_row['name'],
    arrival_time=sample_row['start_slot'],
    cpu_req=sample_row['cpu_milli'],
    gpu_req=sample_row['gpu_milli'],
    gpu_type=sample_row['gpu_spec'],
    duration=sample_row['duration']
)

print(sample_task)
```

```python
class WirelessModel:
    """Calculates transmission rates based on Shannon capacity as per LRMA paper."""
    def __init__(self, bandwidth=20e6, noise_power=1e-13, trans_power=0.5, channel_gain=1e-3):
        self.B = bandwidth        # B
        self.sigma2 = noise_power  # \sigma^2
        self.p_tx = trans_power    # p_n(t)
        self.h = channel_gain      # h_n(t)

    def calculate_rate(self):
        """
        Calculates r_n(t) = B * log2(1 + (p_n(t) * h_n(t)) / sigma^2)
        """
        snr = (self.p_tx * self.h) / self.sigma2
        rate = self.B * np.log2(1 + snr)
        return rate # bit/s

# Initialize wireless environment and calculate transmission rate
wireless = WirelessModel()
r_n_t = wireless.calculate_rate()

print(f"Wireless Transmission Rate (r_n(t)): {r_n_t / 1e6:.2f} Mbps")
```

```python
class MHFQ:
    """Multi-Level Heterogeneous Fairness Queue from LRMA paper."""
    def __init__(self, num_mes):
        # Each MES node has its own set of queues indexed by GPU Type
        self.queues = {i: {} for i in range(num_mes)}

    def enqueue(self, mes_id, task):
        """Push task into the specific queue for its GPU type R."""
        gpu_type = task.R
        if gpu_type not in self.queues[mes_id]:
            self.queues[mes_id][gpu_type] = []
        self.queues[mes_id][gpu_type].append(task)

    def get_queue_length(self, mes_id, gpu_type):
        """Returns Q(t) for a specific GPU resource type."""
        if gpu_type in self.queues[mes_id]:
            return sum(t.size for t in self.queues[mes_id][gpu_type])
        return 0.0

# Initialize MHFQ for the 5 MES nodes defined in EnvConfig
mhfq = MHFQ(num_mes=EnvConfig.NUM_MES)

# Test enqueuing the sample task to MES 0
mhfq.enqueue(mes_id=0, task=sample_task)

print(f"MHFQ Initialized.")
print(f"Current backlog in MES 0 for GPU type '{sample_task.R}': {mhfq.get_queue_length(0, sample_task.R):.2f} MB")
```

```python
class LRMARewardCalculator:
    """Calculates rewards based on the Lyapunov Drift-Plus-Penalty (DPP) method."""
    def __init__(self, V_penalty=10.0):
        self.V = V_penalty

    def calculate_drift(self, current_queues, next_queues):
        """Calculates Delta(t) = 0.5 * sum(Q(t+1)^2 - Q(t)^2)"""
        drift = 0.5 * (np.sum(np.square(next_queues)) - np.sum(np.square(current_queues)))
        return drift

    def calculate_reward(self, drift, energy, delay):
        """
        Reward = -(Drift + V * (Energy + Delay))
        In DRL, we maximize reward, which is equivalent to minimizing Drift + Penalty.
        """
        penalty = self.V * (energy + delay)
        return -(drift + penalty)

# Initialize Reward Calculator
reward_calc = LRMARewardCalculator(V_penalty=EnvConfig.V)

print("Lyapunov Reward Calculator Initialized.")
```

```python
import torch
import torch.nn as nn

class WorkloadPredictor(nn.Module):
    """LSTM-based future workload predictor as specified in LRMA paper."""
    def __init__(self, input_dim=1, hidden_dim=64, num_layers=2):
        super(WorkloadPredictor, self).__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1) # Predicts next slot workload intensity

    def forward(self, x):
        # x shape: (batch, seq_len, input_dim)
        out, _ = self.lstm(x)
        out = self.fc(out[:, -1, :])
        return out

# Initialize the LSTM predictor
predictor = WorkloadPredictor()
print("LSTM Workload Predictor Initialized.")

# Example prediction logic placeholder
# In training, we will feed the 'start_slot' counts from the Alibaba loader into this model.
def get_future_workload_estimate(past_workload_sequence):
    predictor.eval()
    with torch.no_grad():
        seq_tensor = torch.FloatTensor(past_workload_sequence).view(1, -1, 1)
        prediction = predictor(seq_tensor)
        return prediction.item()
```

```python
def prepare_workload_timeseries(df, slot_duration_seconds=60):
    """Aggregates task arrivals into fixed-length time slots for LSTM training."""
    # Calculate which slot each task falls into
    df['slot_id'] = (df['start_slot'] / slot_duration_seconds).astype(int)

    # Count arrivals per slot
    workload_counts = df.groupby('slot_id').size().reset_index(name='task_count')

    # Ensure we have a continuous sequence (fill empty slots with 0)
    full_range = np.arange(workload_counts['slot_id'].min(), workload_counts['slot_id'].max() + 1)
    workload_series = pd.DataFrame({'slot_id': full_range})
    workload_series = workload_series.merge(workload_counts, on='slot_id', how='left').fillna(0)

    return workload_series['task_count'].values

# Generate the time series from the loaded Alibaba pods
raw_workload_data = prepare_workload_timeseries(loader.pods_df)

print(f"Generated workload time series with {len(raw_workload_data)} slots.")
print(f"Average tasks per slot: {raw_workload_data.mean():.2f}")

# Visualize a snippet of the workload
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 4))
plt.plot(raw_workload_data[:200])
plt.title("Alibaba Task Arrival Workload (First 200 Slots)")
plt.xlabel("Time Slot")
plt.ylabel("Number of Tasks")
plt.show()
```

```python
from torch.utils.data import DataLoader, Dataset

class WorkloadDataset(Dataset):
    def __init__(self, data, seq_length=10):
        self.data = torch.FloatTensor(data)
        self.seq_length = seq_length

    def __len__(self):
        return len(self.data) - self.seq_length

    def __getitem__(self, idx):
        return self.data[idx:idx+self.seq_length].view(-1, 1), self.data[idx+self.seq_length]

def train_predictor(model, data, epochs=5, seq_length=10, lr=0.001):
    model.train()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.MSELoss()
    dataset = WorkloadDataset(data, seq_length)
    loader = DataLoader(dataset, batch_size=64, shuffle=True)

    for epoch in range(epochs):
        total_loss = 0
        for x, y in loader:
            optimizer.zero_grad()
            y_pred = model(x)
            loss = criterion(y_pred.squeeze(), y)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        print(f"Epoch {epoch+1}/{epochs}, Loss: {total_loss/len(loader):.6f}")

# Train on a subset of the data for demonstration
train_predictor(predictor, raw_workload_data[:10000])
print("LSTM Predictor training complete.")
```

```python
class DRLActor(nn.Module):
    """Actor network for offloading decisions (ED or Cloud Actors)."""
    def __init__(self, state_dim, action_dim, hidden_dim=128):
        super(DRLActor, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, action_dim),
            nn.Softmax(dim=-1)
        )

    def forward(self, state):
        return self.net(state)

class DRLCritic(nn.Module):
    """Critic network to estimate the value function V(s)."""
    def __init__(self, state_dim, hidden_dim=128):
        super(DRLCritic, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 1)
        )

    def forward(self, state):
        return self.net(state)

# State includes: Local queue, MES queues, Predicted Workload, and Task Requirements
# For LRMA: state = [Q_local, Q_mes_1..N, C, G, R, size, beta_hat]
state_dim = 1 + EnvConfig.NUM_MES + 5
action_dim = EnvConfig.NUM_MES + 1 # Offload to MES 1..N or Local

actor = DRLActor(state_dim, action_dim)
critic = DRLCritic(state_dim)

print(f"DRL Agent Initialized: State Dim = {state_dim}, Action Dim = {action_dim}")
```

```python
class LRMAExperienceReplay:
    """Buffer to store trajectories for DRL training."""
    def __init__(self):
        self.states = []
        self.actions = []
        self.rewards = []
        self.next_states = []
        self.log_probs = []

    def add(self, state, action, reward, next_state, log_prob):
        self.states.append(state)
        self.actions.append(action)
        self.rewards.append(reward)
        self.next_states.append(next_state)
        self.log_probs.append(log_prob)

    def clear(self):
        self.states = []
        self.actions = []
        self.rewards = []
        self.next_states = []
        self.log_probs = []

# Instantiate the buffer
buffer = LRMAExperienceReplay()

def select_action(actor_net, state):
    """Selects an action based on the actor's probability distribution."""
    state_t = torch.FloatTensor(state).unsqueeze(0)
    probs = actor_net(state_t)
    dist = torch.distributions.Categorical(probs)
    action = dist.sample()
    return action.item(), dist.log_prob(action)

print("Experience Replay Buffer and Action Selection logic ready.")
```

```python
def train_drl_agent(actor, critic, buffer, optimizer_a, optimizer_c, gamma=0.99, eps_clip=0.2):
    """Performs a PPO-style update on the DRL agent networks."""
    if len(buffer.states) == 0: return

    # Convert buffer lists to tensors
    old_states = torch.FloatTensor(np.array(buffer.states))
    old_actions = torch.LongTensor(np.array(buffer.actions))
    old_log_probs = torch.stack(buffer.log_probs).detach()
    rewards = torch.FloatTensor(np.array(buffer.rewards))

    # Calculate advantages and targets
    with torch.no_grad():
        state_values = critic(old_states).squeeze()
        # Simple advantage calculation
        advantages = rewards - state_values
        # Normalize advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

    # PPO Update for Actor
    probs = actor(old_states)
    dist = torch.distributions.Categorical(probs)
    new_log_probs = dist.log_prob(old_actions)

    ratios = torch.exp(new_log_probs - old_log_probs)
    surr1 = ratios * advantages
    surr2 = torch.clamp(ratios, 1-eps_clip, 1+eps_clip) * advantages
    actor_loss = -torch.min(surr1, surr2).mean()

    optimizer_a.zero_grad()
    actor_loss.backward()
    optimizer_a.step()

    # Update Critic
    current_values = critic(old_states).squeeze()
    critic_loss = nn.MSELoss()(current_values, rewards)

    optimizer_c.zero_grad()
    critic_loss.backward()
    optimizer_c.step()

    buffer.clear()
    return actor_loss.item(), critic_loss.item()

# Setup optimizers
optimizer_actor = torch.optim.Adam(actor.parameters(), lr=0.0003)
optimizer_critic = torch.optim.Adam(critic.parameters(), lr=0.001)

print("DRL Training Step function and optimizers configured.")
```

```python
class LRMA_Environment:
    def __init__(self, loader, predictor, config):
        self.loader = loader
        self.predictor = predictor
        self.config = config
        self.mhfq = MHFQ(config.NUM_MES)
        self.wireless = WirelessModel()
        self.reward_calc = LRMARewardCalculator(V_penalty=config.V)

        self.current_time = 0
        self.slot_duration = 60 # seconds
        self.history_workload = []

    def get_state(self, task):
        # 1. Local backlog (simplified for simulation as 0 or task size)
        q_local = task.size

        # 2. MES backlogs for the required GPU type R
        q_mes = [self.mhfq.get_queue_length(i, task.R) for i in range(self.config.NUM_MES)]

        # 3. Task features
        task_features = [task.C, task.G, (1 if task.R != 'MISC' else 0), task.size]

        # 4. Predicted Workload
        beta_hat = get_future_workload_estimate(self.history_workload[-10:] if len(self.history_workload) >= 10 else [0]*10)

        state = np.array([q_local] + q_mes + task_features + [beta_hat])
        return state

    def step(self, task, action):
        # Action 0: Local, Actions 1-N: MES Nodes
        # In a real step, we would calculate energy and delay
        # For now, we simulate a dummy penalty based on queue length and distance

        if action == 0:
            energy, delay = 0.5 * task.size, 1.0 * task.size
        else:
            mes_idx = action - 1
            self.mhfq.enqueue(mes_idx, task)
            # Offloading delay includes transmission time
            trans_time = task.size / (self.wireless.calculate_rate() / 8e6) # size in MB, rate in Mbps
            energy, delay = 0.2 * task.size, trans_time + (self.mhfq.get_queue_length(mes_idx, task.R) / 10.0)

        return energy, delay

# Instantiate the environment
env = LRMA_Environment(loader, predictor, EnvConfig)
print("LRMA Environment Orchestrator Initialized.")
```

```python
def run_lrma_training(num_tasks=500, update_interval=64):
    """Main simulation and training loop for the LRMA system."""
    print(f"Starting training for {num_tasks} tasks...")

    # Get a batch of tasks from the loader
    tasks_df = loader.pods_df.head(num_tasks)

    total_rewards = []
    episode_reward = 0

    for i, row in tasks_df.iterrows():
        # 1. Create task object
        task = LRMATask(row['name'], row['start_slot'], row['cpu_milli'],
                        row['gpu_milli'], row['gpu_spec'], row['duration'])

        # 2. Get state and select action
        state = env.get_state(task)
        action, log_prob = select_action(actor, state)

        # 3. Simulate step and get reward
        q_before = [env.mhfq.get_queue_length(j, task.R) for j in range(EnvConfig.NUM_MES)]
        energy, delay = env.step(task, action)
        q_after = [env.mhfq.get_queue_length(j, task.R) for j in range(EnvConfig.NUM_MES)]

        drift = reward_calc.calculate_drift(np.array(q_before), np.array(q_after))
        reward = reward_calc.calculate_reward(drift, energy, delay)

        # 4. Store experience
        next_state = env.get_state(task) # Simplified next state
        buffer.add(state, action, reward, next_state, log_prob)

        episode_reward += reward
        env.history_workload.append(1) # Record arrival for LSTM

        # 5. Periodic update
        if (i + 1) % update_interval == 0:
            a_loss, c_loss = train_drl_agent(actor, critic, buffer, optimizer_actor, optimizer_critic)
            total_rewards.append(episode_reward)
            print(f"Task {i+1}/{num_tasks} | Reward: {episode_reward:.2f} | Actor Loss: {a_loss:.4f}")
            episode_reward = 0

    return total_rewards

# Run a small training session
training_history = run_lrma_training()

plt.figure(figsize=(8, 4))
plt.plot(training_history)
plt.title("DRL Training Convergence (Lyapunov Reward)")
plt.xlabel("Update Interval")
plt.ylabel("Cumulative Reward")
plt.show()
```

### Full-Scale Simulation (8,000+ Tasks)
We now execute the LRMA training loop over the entire Alibaba dataset to observe long-term stability and performance.

```python
import time

# Reset environment for a clean full run
env = LRMA_Environment(loader, predictor, EnvConfig)
buffer.clear()

def run_full_lrma_simulation(update_interval=64):
    all_tasks = loader.pods_df
    num_total = len(all_tasks)
    print(f"Starting full simulation for {num_total} tasks...")

    full_history = []
    episode_reward = 0
    start_time = time.time()

    for i, row in all_tasks.iterrows():
        # 1. Task Creation
        task = LRMATask(row['name'], row['start_slot'], row['cpu_milli'],
                        row['gpu_milli'], row['gpu_spec'], row['duration'])

        # 2. State & Action
        state = env.get_state(task)
        action, log_prob = select_action(actor, state)

        # 3. Environment Step
        q_before = [env.mhfq.get_queue_length(j, task.R) for j in range(EnvConfig.NUM_MES)]
        energy, delay = env.step(task, action)
        q_after = [env.mhfq.get_queue_length(j, task.R) for j in range(EnvConfig.NUM_MES)]

        drift = reward_calc.calculate_drift(np.array(q_before), np.array(q_after))
        reward = reward_calc.calculate_reward(drift, energy, delay)

        # 4. Experience Logging
        next_state = env.get_state(task)
        buffer.add(state, action, reward, next_state, log_prob)

        episode_reward += reward
        env.history_workload.append(1)

        # 5. Network Updates
        if (i + 1) % update_interval == 0:
            train_drl_agent(actor, critic, buffer, optimizer_actor, optimizer_critic)
            full_history.append(episode_reward)
            episode_reward = 0

        if (i + 1) % 500 == 0:
            print(f"Progress: {i+1}/{num_total} tasks processed...")

    end_time = time.time()
    print(f"Full simulation complete in {end_time - start_time:.2f} seconds.")
    return full_history

full_training_history = run_full_lrma_simulation()
```

### Tuning the Lyapunov Parameter $V$
We will test $V \in \{1, 50, 100\}$ to observe how the agent prioritizes queue stability versus processing efficiency.

```python
def train_with_v(v_value, num_tasks=2000):
    # Re-initialize configuration and networks
    config = EnvConfig()
    config.V = v_value

    v_actor = DRLActor(state_dim, action_dim)
    v_critic = DRLCritic(state_dim)
    v_opt_a = torch.optim.Adam(v_actor.parameters(), lr=0.0003)
    v_opt_c = torch.optim.Adam(v_critic.parameters(), lr=0.001)

    v_env = LRMA_Environment(loader, predictor, config)
    v_reward_calc = LRMARewardCalculator(V_penalty=v_value)
    v_buffer = LRMAExperienceReplay()

    history = []
    episode_reward = 0
    tasks_df = loader.pods_df.head(num_tasks)

    for i, row in tasks_df.iterrows():
        task = LRMATask(row['name'], row['start_slot'], row['cpu_milli'],
                        row['gpu_milli'], row['gpu_spec'], row['duration'])

        state = v_env.get_state(task)
        action, log_prob = select_action(v_actor, state)

        q_before = [v_env.mhfq.get_queue_length(j, task.R) for j in range(config.NUM_MES)]
        energy, delay = v_env.step(task, action)
        q_after = [v_env.mhfq.get_queue_length(j, task.R) for j in range(config.NUM_MES)]

        drift = v_reward_calc.calculate_drift(np.array(q_before), np.array(q_after))
        reward = v_reward_calc.calculate_reward(drift, energy, delay)

        v_buffer.add(state, action, reward, v_env.get_state(task), log_prob)
        episode_reward += reward

        if (i + 1) % 64 == 0:
            train_drl_agent(v_actor, v_critic, v_buffer, v_opt_a, v_opt_c)
            history.append(episode_reward)
            episode_reward = 0

    return history

print("Training with V=1...")
hist_v1 = train_with_v(1.0)
print("Training with V=100...")
hist_v100 = train_with_v(100.0)
```

### Training Convergence Time Analysis
We define 'convergence' as the point where the reward stays within a certain percentage of its final moving average. This allows us to quantify which Lyapunov setting allows the DRL agent to learn the optimal policy faster.

```python
import matplotlib.pyplot as plt
import numpy as np

def calculate_convergence_interval(history, threshold=0.05):
    """Finds the first interval where reward stabilizes within threshold of final mean."""
    final_mean = np.mean(history[-10:])
    for i, r in enumerate(history):
        if abs(r - final_mean) < abs(threshold * final_mean):
            return i
    return len(history)

# Calculate convergence points
conv_v1 = calculate_convergence_interval(hist_v1)
conv_v10 = calculate_convergence_interval(full_training_history[:len(hist_v1)])
conv_v100 = calculate_convergence_interval(hist_v100)

# Visualization
plt.figure(figsize=(12, 6))
plt.plot(hist_v1, label=f'V=1 (Conv at {conv_v1})', alpha=0.8)
plt.plot(full_training_history[:len(hist_v1)], label=f'V=10 (Conv at {conv_v10})', alpha=0.8)
plt.plot(hist_v100, label=f'V=100 (Conv at {conv_v100})', alpha=0.8)

plt.axvline(conv_v1, color='blue', linestyle='--', alpha=0.5)
plt.axvline(conv_v10, color='orange', linestyle='--', alpha=0.5)
plt.axvline(conv_v100, color='green', linestyle='--', alpha=0.5)

plt.title('Convergence Speed Comparison Across V Parameters')
plt.xlabel('Training Intervals (64 tasks per interval)')
plt.ylabel('Interval Reward')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Convergence Intervals (Lower is faster):\n- V=1: {conv_v1} intervals\n- V=10: {conv_v10} intervals\n- V=100: {conv_v100} intervals")
```
